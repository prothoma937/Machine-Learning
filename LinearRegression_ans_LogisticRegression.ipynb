```python
#Import Libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
#Create a Simple Dataset
hours = np.array([2, 4, 6, 8, 10]).reshape(-1,1)
marks = np.array([55, 60, 70, 80, 88])

hours, marks
```

```python
#train liner mode
model= LinearRegression()
model.fit(hours,marks)
```

```python
#view the learnined parameter
print('m',model.coef_[0])
print('c', model.intercept_)
```

```python
#visualization
plt.scatter(hours, marks, color='blue', label='Best fit line')
predicted_marks=model.predict(hours)
plt.plot(hours, predicted_marks, color='red', label='Best fit line')
```

```python
new_hours=np.array([[7]])
new_p=model.predict(new_hours)
new_p
```

```python
from sklearn.metrics import mean_squared_error
mse=mean_squared_error(marks, predicted_marks)
mse
```

```python
#Use a Gradient-Descent-Based Model – SGDRegressor
from sklearn.linear_model import LinearRegression, SGDRegressor


sgd_r=SGDRegressor(
    max_iter=1000,
    learning_rate='invscaling',
    eta0=0.01,
    random_state=42

)

sgd_r.fit(hours, marks)
```

```python
print("m: ", sgd_r.coef_[0])
print("c: ",sgd_r.intercept_[0])


predicted_sgd=sgd_r.predict(hours)
mse_sgd=mean_squared_error(marks, predicted_marks)
mse_sgd
```

```python
plt.scatter(hours,marks, color='blue')

plt.plot(hours, predicted_marks, label='linerregression', linestyle='--')
plt.plot(hours, predicted_sgd, label='linerregression', linestyle='-.')
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
#Compute Metrics
mae = mean_absolute_error(marks,predicted_marks)
mse = mean_squared_error(marks, predicted_marks)
rmse = np.sqrt(mse)
r2 = r2_score(marks, predicted_marks)

print("MAE: ", mae)
print("MSE", mse)
print("RMSE", rmse)
print("R^2:",r2)
```

```python
#Visualizing the Fit
plt.scatter(hours, marks, label="Actual Data")
plt.plot(hours, predicted_marks, color="red", label="Predicted Line")
```
