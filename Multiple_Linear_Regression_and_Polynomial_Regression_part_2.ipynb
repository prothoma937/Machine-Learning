```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import fetch_california_housing
plt.rcParams['figure.figsize']=(8,6)
plt.rcParams['axes.grid']=True
```

```python
california= fetch_california_housing( as_frame=True)
df= california.frame.copy()
df.head()
```

```python
# Define target and feature columns

target_col ='MedHouseVal'

feature_cols = ['MedInc', 'HouseAge', 'AveRooms', 'AveBedrms', 'Population', 'AveOccup']
```

```python
corr_matrix=df [feature_cols +[target_col]].corr()
corr_matrix
```

```python
fig, ax= plt.subplots()
cax=ax.imshow(corr_matrix, interpolation='nearest')
ax.set_xticks(range(len(corr_matrix.columns)))
ax.set_yticks(range(len(corr_matrix.index)))
ax.set_xticklabels(corr_matrix.columns, rotation=45, ha='right')
ax.set_yticklabels(corr_matrix.index)
fig.colorbar(cax)
ax.set_title('Correlation matrix for selected features and target')
plt.tight_layout()
plt.show()
```

```python
#trrain and test split
from sklearn.model_selection import train_test_split
x=df[feature_cols]
y=df[target_col]
x_train, x_test, y_train, y_test = train_test_split( x,y, test_size=0.2, random_state=42)
print('Training set size', x_train.shape[0],'rows')
print('Test set size', x_test.shape[0],'rows')
```

```python
#create and fit the linear regresssion mode
from sklearn.linear_model import LinearRegression
lin_reg= LinearRegression()
lin_reg.fit(x_train, y_train)
print('model fitted successfully')
```

```python
print('intercept(bias):',lin_reg.intercept_)
print('\n cofficients:')
for feature_name, coef in zip( feature_cols, lin_reg.coef_):
  print(f'{feature_name}:{coef}')
```

```python
#make a predictions on training and test sets
y_train_pred= lin_reg.predict(x_train)
y_test_pred= lin_reg.predict(x_test)

print('some sample prediction on test set')
print('Predicted:', y_test_pred[:5])
print('Actual:', y_test.values[:5])
```

```python
y_train_pred=lin_reg.predict(x_train)
y_test_pred=lin_reg.predict(x_test)
print('some sample predicted  on test set')
print('predicted:', y_test_pred[:5])
print('predicted:', y_test.values[:5])
```

```python
# Step 5: Define a helper function to print evaluation metrics
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
def regresssion_metrix(y_true, y_pred, label='Model'):
  mae=mean_absolute_error(y_true, y_pred)
  mse=mean_squared_error(y_true, y_pred)
  rmse=np.sqrt(mse)
  r2=r2_score(y_true, y_pred)

  print(f'== {label} ==')
  print('MAE: ', mae)
  print('RMSE: ', rmse)
  print('R2 : ', r2)
  print()

regresssion_metrix(y_train, y_train_pred, label='Linear Regression(train)')
regresssion_metrix(y_test, y_test_pred, label='liner Reression(Test)')
```

```python
from matplotlib.lines import lineStyles
# Plot predicted vs actual values on the test set
import matplotlib.pyplot as plt

plt.figure()
plt.scatter(y_test, y_test_pred, alpha=0.3, )
plt.xlabel('Actual Value')
plt.ylabel('Predicted Value')

#diagonal referenceline
min_val= min(y_test.min(), y_test_pred.min())
max_val= max(y_test.max(), y_test_pred.max())
plt.plot([min_val, max_val], [min_val, max_val], linestyle='--')
```

```python
# Residual plot (errors = actual - predicted)

residual= y_test - y_test_pred
plt.figure()
plt.scatter(y_test_pred, residual, alpha=0.3)
plt.axhline(0, linestyle='--')
```
